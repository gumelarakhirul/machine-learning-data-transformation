## 1. Import Library

```python
import pandas as pd
import numpy as np

from sklearn.preprocessing import LabelEncoder
from sklearn.preprocessing import MinMaxScaler
from sklearn.preprocessing import StandardScaler
```

## 2. Membuat Dataset ke dalam DataFrame `df`

```python
data = {
    "Jam_Belajar": [23, 19, 24, 29, 19, 19, 29, 25, 17, 23,
                    17, 17, 21, 9, 10, 17, 14, 22, 15, 12,
                    29, 19, 20, 11, 17, 21, 13, 22, 16, 18],

    "Kehadiran": [84, 64, 98, 89, 92, 88, 84, 78, 94, 98,
                  80, 97, 83, 82, 78, 68, 60, 70, 80, 75,
                  78, 99, 74, 78, 65, 62, 91, 83, 90, 66],

    "Ekstrakulikuler": ["Tidak", "Tidak", "Iya", "Iya", "Iya", "Iya", "Iya", "Iya", "Tidak", "Iya",
                        "Tidak", "Iya", "Iya", "Iya", "Iya", "Tidak", "Iya", "Iya", "Iya", "Iya",
                        "Tidak", "Tidak", "Tidak", "Iya", "Iya", "Iya", "Tidak", "Iya", "Tidak", "Tidak"],

    "Jam_Tidur": [7, 8, np.nan, 8, 6, 8, 7, np.nan, 6, np.nan,
                  8, 6, 8, 8, 8, 8, 10, np.nan, 9, np.nan,
                  5, 6, np.nan, 8, 5, 6, 6, 6, 4, 4],

    "Sesi_Les": [0, 2, 2, 1, 3, 3, 1, 1, 0, 0,
                 4, 2, 2, 2, 1, 2, 0, 1, 3, 3,
                 0, 1, 1, 1, 2, 0, 1, 0, 1, 2],

    "Jarak_Dari_Rumah": ["Dekat", "Sedang", "Dekat", "Sedang", "Dekat", "Dekat", "Sedang", "Jauh", "Dekat", "Sedang",
                         "Sedang", "Dekat", "Dekat", "Dekat", "Dekat", "Dekat", "Dekat", "Dekat", "Sedang", "Dekat",
                         "Sedang", "Dekat", "Sedang", "Sedang", "Dekat", "Jauh", "Sedang", "Sedang", "Sedang", "Dekat"],

    "Jenis_kelamin": ["Laki-laki", "Perempuan", "Laki-laki", "Laki-laki", "Perempuan", "Laki-laki", "Laki-laki", "Laki-laki", "Laki-laki", "Laki-laki",
                      "Laki-laki", "Laki-laki", "Laki-laki", "Laki-laki", "Laki-laki", "Perempuan", "Laki-laki", "Perempuan", "Perempuan", "Laki-laki",
                      "Perempuan", "Laki-laki", "Perempuan", "Laki-laki", "Perempuan", "Laki-laki", "Perempuan", "Laki-laki", "Perempuan", "Laki-laki"],

    "Nilai_Ujian": [67, 61, 74, 71, 70, 71, 67, 66, 69, 72,
                    68, 71, 70, 66, 65, 64, 60, 65, 67, 66,
                    69, 72, 66, 66, 63, 64, 65, 71, 66, 64]
}

df = pd.DataFrame(data)

df
```

## 3. Cek Informasi Dataset

```python
print("Jumlah baris dan kolom:", df.shape)
print("\nInformasi dataset:")
df.info()
```

```python
print("Missing value sebelum ditangani:")
print(df.isnull().sum())
```

## 4.1 Handling Missing Values - A. Imputation

Pada bagian ini, nilai kosong pada kolom `Jam_Tidur` diisi menggunakan nilai rata-rata atau `mean()`.

```python
df_imputation = df.copy()

rata_rata_jam_tidur = df_imputation['Jam_Tidur'].mean()
print("Rata-rata Jam_Tidur:", rata_rata_jam_tidur)

df_imputation['Jam_Tidur'] = df_imputation['Jam_Tidur'].fillna(rata_rata_jam_tidur)

print("\nMissing value setelah imputation:")
print(df_imputation.isnull().sum())

df_imputation.head(10)
```

## 4.1 Handling Missing Values - B. Deletion

Pada bagian ini, baris yang memiliki nilai kosong pada kolom `Jam_Tidur` dihapus menggunakan `dropna()`.

```python
df_deletion = df.copy()

print("Jumlah data sebelum dropna:", len(df_deletion))

df_deletion.dropna(subset=['Jam_Tidur'], how='any', inplace=True)

print("Jumlah data setelah dropna:", len(df_deletion))
print("\nMissing value setelah deletion/dropna:")
print(df_deletion.isnull().sum())

df_deletion.head(10)
```

## Catatan

Untuk tahap berikutnya digunakan `df_imputation`, karena data tetap lengkap 30 baris dan missing value sudah diisi dengan rata-rata.

```python
df_encoded = df_imputation.copy()
df_encoded.head()
```

## 4.2 Encoding - A. Label Encoding untuk `Ekstrakulikuler`

Fitur `Ekstrakulikuler` diubah secara manual:
- `Iya` menjadi `1`
- `Tidak` menjadi `0`

```python
df_encoded['Ekstrakulikuler'] = df_encoded['Ekstrakulikuler'].apply(
    lambda x: 1 if x == 'Iya' else (0 if x == 'Tidak' else None)
)

df_encoded[['Ekstrakulikuler']].head(10)
```

## 4.2 Encoding - A. Label Encoding untuk `Jenis_kelamin`

Fitur `Jenis_kelamin` diubah menggunakan `LabelEncoder` dari library `sklearn`.

```python
le = LabelEncoder()

df_encoded['Jenis_kelamin'] = le.fit_transform(df_encoded['Jenis_kelamin'])

print("Kelas pada LabelEncoder:", le.classes_)
df_encoded[['Jenis_kelamin']].head(10)
```

### Jawaban Pertanyaan

Perbedaan kode encoding pada fitur `Ekstrakulikuler` dan `Jenis_kelamin` adalah sebagai berikut:

1. Encoding pada fitur `Ekstrakulikuler` dilakukan secara manual menggunakan fungsi `apply()` dan `lambda`. Nilai `Iya` ditentukan menjadi `1`, sedangkan nilai `Tidak` ditentukan menjadi `0`.
2. Encoding pada fitur `Jenis_kelamin` dilakukan menggunakan `LabelEncoder` dari library `sklearn`. Proses perubahan kategori menjadi angka dilakukan otomatis berdasarkan kelas/kategori yang ditemukan pada kolom tersebut.

Jadi, perbedaannya adalah encoding `Ekstrakulikuler` nilainya ditentukan sendiri secara manual, sedangkan encoding `Jenis_kelamin` dilakukan otomatis oleh `LabelEncoder`.

## 4.2 Encoding - B. One-Hot Encoding untuk `Jarak_Dari_Rumah`

Fitur `Jarak_Dari_Rumah` diubah menjadi beberapa kolom biner menggunakan `pd.get_dummies()`.

```python
df_encoded = pd.get_dummies(
    df_encoded,
    prefix=['Jarak_Rumah'],
    columns=['Jarak_Dari_Rumah'],
    dtype=int
)

df_encoded.head(10)
```

## 4.3 Scaling - A. Normalization

Normalization menggunakan `MinMaxScaler`, sehingga nilai fitur berada pada rentang 0 sampai 1.

```python
scaler = MinMaxScaler()

df_normalized = pd.DataFrame(
    scaler.fit_transform(df_encoded),
    columns=df_encoded.columns
)

df_normalized.head(10)
```

## 4.3 Scaling - B. Standardization

Standardization menggunakan `StandardScaler`, sehingga data memiliki rata-rata mendekati 0 dan standar deviasi mendekati 1. Data yang digunakan adalah data hasil encoding, bukan data hasil normalisasi.

```python
scaler = StandardScaler()

scaled_data = scaler.fit_transform(df_encoded)

df_standardized = pd.DataFrame(
    scaled_data,
    columns=df_encoded.columns
)

df_standardized.head(10)
```

## 4.4 Creating New Feature

Pada bagian ini dibuat fitur baru dari fitur yang sudah ada. Contoh fitur baru yang dibuat adalah `Produktivitas_Belajar`, yaitu hasil perkalian antara `Jam_Belajar` dan `Kehadiran`.

```python
df_feature = df_encoded.copy()

df_feature['Produktivitas_Belajar'] = df_feature['Jam_Belajar'] * df_feature['Kehadiran']

df_feature[['Jam_Belajar', 'Kehadiran', 'Produktivitas_Belajar']].head(10)
```

## Contoh Creating New Feature Lain Sesuai Modul

Contoh berikut menghapus satuan `kg` dari data berat, lalu membuat kolom baru `berat_dalam_kg`.

```python
data_berat = pd.DataFrame({
    "berat": ["10 kg", "20 kg", "18 kg", "25 kg", "19 kg", "21 kg"]
})

data_berat["berat_dalam_kg"] = data_berat["berat"].replace(" kg", "", regex=True).astype(float)

data_berat
```

## Kesimpulan Percobaan

Berdasarkan percobaan yang dilakukan, proses transformasi data terdiri dari beberapa tahap penting. Missing value pada kolom `Jam_Tidur` dapat ditangani dengan dua cara, yaitu imputation menggunakan rata-rata dan deletion menggunakan `dropna()`. Data kategorikal seperti `Ekstrakulikuler`, `Jenis_kelamin`, dan `Jarak_Dari_Rumah` perlu diubah menjadi numerik melalui Label Encoding dan One-Hot Encoding. Setelah data menjadi numerik, scaling dapat dilakukan menggunakan Normalization dan Standardization. Selain itu, fitur baru juga dapat dibuat dari fitur yang sudah ada untuk menambah informasi pada dataset.
